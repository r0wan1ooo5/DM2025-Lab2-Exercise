### 1.2 Feature Engineering Steps

Two parallel feature engineering strategies were applied to better capture both lexical patterns and emotional signals from the text:

* TF-IDF Vectorization — Used to convert raw text into high-dimensional sparse vectors that represent word importance based on term frequency and inverse document frequency. This helps the model learn from keyword usage, phrase patterns, and contextual expressions.

* Sentiment Feature Extraction — Using a pretrained sentiment analysis model, each text was also annotated with its polarity (negative, neutral, positive) and corresponding confidence scores. These were numerically encoded and appended as additional features, allowing the model to incorporate emotional intensity beyond word frequency.

These two types of features are independent and complementary, providing both linguistic (TF-IDF) and affective (sentiment) representations for downstream classification.

### 1.3 Explanation of Your Model


I adopted an ensemble learning strategy using four different machine learning models: Support Vector Machine (SVM), Logistic Regression (LR), Random Forest, and XGBoost.
Each model independently predicted the emotion label, and their outputs were combined using a voting mechanism:

* If a majority (≥2 models) agreed on the same prediction, that label was selected.

* If all four predictions were different, or a 2 vs. 2 tie occurred, the system selected the prediction from the model with the highest validation F1-score, ensuring better reliability.

This ensemble approach helps leverage the strengths of different models and improves robustness and overall classification performance.

```python
import re
import nltk
import string


from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer


nltk.download('stopwords')
nltk.download('wordnet')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    # change to lower case
    text = text.lower()

    # remove URLs / emails
    text = re.sub(r'http\S+|www\S+|https\S+|@\S+', '', text)

    # remove HTML tag
    text = re.sub(r'<.*?>', '', text)


    # remove numbers and punctuation (keep letters and spaces)
    text = re.sub(f'[^a-zA-Z\s]', '', text)

    # remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # remove stop words & lemmatization
    cleaned_words = []
    for word in text.split():
        if word not in stop_words:
            cleaned_words.append(lemmatizer.lemmatize(word))

    return " ".join(cleaned_words)


df_posts['clean_text'] = df_posts['text'].apply(clean_text)

print(df_posts[['text', 'clean_text']].head(10))
```

```python
import torch
from transformers import pipeline

pipe = pipeline(
    "text-classification",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest"
)

```

```python
import math
from tqdm import tqdm

def add_sentiment_roberta(df, text_col="clean_text", batch_size=32):
    """
    使用 cardiffnlp/twitter-roberta-base-sentiment-latest
    為文字資料添加情緒極性標記 (negative / neutral / positive)

    新增欄位：
    - sentiment_label
    - sentiment_score
    - sentiment_score_num (轉成 0/1/2)

    Using cardiffnlp/twitter-roberta-base-sentiment-latest
    to assign sentiment polarity labels (negative / neutral / positive) to text data.

    New columns added:
    - sentiment_label
    - sentiment_score
    - sentiment_score_num (converted to 0 / 1 / 2)
    """
    texts = df[text_col].fillna("").tolist()
    all_labels = []
    all_scores = []

    n = len(texts)
    num_batches = math.ceil(n / batch_size)

    for i in tqdm(range(num_batches), desc="Generating sentiment with RoBERTa"):
        batch_texts = texts[i * batch_size : (i + 1) * batch_size]
        outputs = pipe(batch_texts, truncation=True, max_length=256)

        for out in outputs:
            all_labels.append(out["label"])
            all_scores.append(out["score"])

    df = df.copy()
    df["sentiment_label"] = all_labels
    df["sentiment_score"] = all_scores

    # translate text label to number
    label_map = {"negative": 0, "neutral": 1, "positive": 2}
    df["sentiment_score_num"] = df["sentiment_label"].map(label_map)

    return df
```

```python
from sklearn.model_selection import train_test_split

X_text = df_train_sent["clean_text"].fillna("")
y = df_train_sent["emotion_id"].astype(int)  # int labels
# sentiment score 
sent_all = df_train_sent["sentiment_score_num"].fillna(1)  

# train-val split
X_tr_text, X_val_text, y_tr, y_val, sent_tr, sent_val = train_test_split(
    X_text,
    y,
    sent_all,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Train size:", X_tr_text.shape)
print("Val size:", X_val_text.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack

# TF-IDF
tfidf = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2)   # uni + bi-gram
)
X_tr_tfidf = tfidf.fit_transform(X_tr_text)
X_val_tfidf = tfidf.transform(X_val_text)


sent_num_tr = df_train_sent.loc[X_tr_text.index, "sentiment_score_num"].fillna(1)
sent_score_tr = df_train_sent.loc[X_tr_text.index, "sentiment_score"].fillna(0.5)

sent_num_val = df_train_sent.loc[X_val_text.index, "sentiment_score_num"].fillna(1)
sent_score_val = df_train_sent.loc[X_val_text.index, "sentiment_score"].fillna(0.5)

# TF-IDF + sentiment_score_num + sentiment_score
X_tr = hstack([X_tr_tfidf, sent_num_tr.values.reshape(-1, 1), sent_score_tr.values.reshape(-1, 1)])
X_val = hstack([X_val_tfidf, sent_num_val.values.reshape(-1, 1), sent_score_val.values.reshape(-1, 1)])

print("Train final feature shape:", X_tr.shape)
print("Val final feature shape:", X_val.shape)


```

```python
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import f1_score, classification_report

# ---- (1) Linear SVM ----
svm_clf = LinearSVC(
    class_weight="balanced",
    random_state=42
)
svm_clf.fit(X_tr, y_tr)
y_val_pred_svm = svm_clf.predict(X_val)
f1_svm = f1_score(y_val, y_val_pred_svm, average="macro")
print("\n🔹 Linear SVM - Val Macro F1:", f1_svm)
print(classification_report(y_val, y_val_pred_svm))

# ---- (2) Logistic Regression ----
lr_clf = LogisticRegression(
    penalty="l2",
    C=1.0,
    class_weight="balanced",
    solver="saga",
    multi_class="multinomial",
    max_iter=200,
    n_jobs=-1
)
lr_clf.fit(X_tr, y_tr)
y_val_pred_lr = lr_clf.predict(X_val)
f1_lr = f1_score(y_val, y_val_pred_lr, average="macro")
print("\n🔹 Logistic Regression - Val Macro F1:", f1_lr)
print(classification_report(y_val, y_val_pred_lr))

# ---- (3) Random Forest ----
rf_clf = RandomForestClassifier(
    n_estimators=400,
    max_depth=35,
    min_samples_leaf=2,
    class_weight="balanced_subsample",
    random_state=42,
    n_jobs=-1
)
rf_clf.fit(X_tr, y_tr)

y_val_pred_rf = rf_clf.predict(X_val)
f1_rf = f1_score(y_val, y_val_pred_rf, average='macro')
print("\n🔹 Random Forest Macro F1:", f1_rf)
print(classification_report(y_val, y_val_pred_rf))

# ---- (4) XGBoost ----
xgb_clf = XGBClassifier(
    objective="multi:softprob",             
    num_class=df_train_sent['emotion'].nunique(),
    eval_metric="mlogloss",
    learning_rate=0.1,
    max_depth=8,
    n_estimators=300,
    subsample=0.8,
    colsample_bytree=0.8,
    tree_method="hist",                       
    random_state=42,
    n_jobs=-1
)

xgb_clf.fit(X_tr, y_tr)
y_val_pred_xgb = xgb_clf.predict(X_val)
f1_xgb = f1_score(y_val, y_val_pred_xgb, average="macro")
print("\n🔹 XGBoost - Val Macro F1:", f1_xgb)
print(classification_report(y_val, y_val_pred_xgb))
```

```python
import numpy as np

def ensemble_vote_4(pred_svm, pred_lr, pred_rf, pred_xgb,
                    f1_svm, f1_lr, f1_rf, f1_xgb):
    """
    四模型投票：
      - 若有多數（≥2票）直接採用
      - 若 4 個模型都不同 → 選驗證 F1 分數最高的那個模型的預測
      - 若出現 2 vs 2 → 選出兩票那個，但若兩組平手 → 再依 F1 高度決定

    Four-model voting strategy:
    - If a majority exists (i.e., at least 2 votes for the same label), choose that label.
    - If all four models predict different labels, choose the prediction from the model with the highest validation F1 score.
    - If a 2 vs 2 tie occurs, select the label supported by 2 votes. 
        If both pairs are equally strong, fall back to the model with the highest F1 score.
    """

    final_pred = []

    f1_scores = {
        "svm": f1_svm,
        "lr": f1_lr,
        "rf": f1_rf,
        "xgb": f1_xgb
    }

    best_model = max(f1_scores, key=f1_scores.get)

    for a, b, c, d in zip(pred_svm, pred_lr, pred_rf, pred_xgb):
        votes = [a, b, c, d]
        labels, counts = np.unique(votes, return_counts=True)
        max_count = counts.max()

        if max_count >= 2:
            final_pred.append(labels[counts.argmax()])
        else:
            if best_model == "svm":
                final_pred.append(a)
            elif best_model == "lr":
                final_pred.append(b)
            elif best_model == "rf":
                final_pred.append(c)
            else:
                final_pred.append(d)

    return np.array(final_pred)

y_val_pred_ens4 = ensemble_vote_4(
    y_val_pred_svm,
    y_val_pred_lr,
    y_val_pred_rf,
    y_val_pred_xgb,
    f1_svm,
    f1_lr,
    f1_rf,
    f1_xgb
)

print("\n🌟 Ensemble (SVM + LR + RF + XGB)  Macro F1:",
      f1_score(y_val, y_val_pred_ens4, average="macro"))
print(classification_report(y_val, y_val_pred_ens4))

```
